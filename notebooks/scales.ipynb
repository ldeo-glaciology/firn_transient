```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
Ec = 60e3
R = 8.314
# Tavg = 263
```

```python
T = np.arange(253, 274, 1)
T
```

```python
Tavg = T.mean()

lam_c = Ec/R/Tavg
G = 50e-3
kappa0 = 2.1
z0 = 50
Tb0 = 10#G*z0/kappa0
tau = Tb0/Tavg
Tavg
```

```python
Tb = T-Tavg
f1 = np.exp(-Ec/R/T)
f2 = np.exp(-Ec/R/(Tb+Tavg))
f1[5], f2[5]
```

```python
f3= np.exp(-Ec/R/Tavg*(2/(Tb/Tavg+1)))
f1[5], f3[5]
```

```python
Tbstar = Tb/Tb0
f4= np.exp(-Ec/R/Tavg*(1/(Tb0*Tbstar/Tavg+1)))
f1[5], f4[5]
```

```python
lam_c = Ec/R/Tavg
tau = Tb0/Tavg
f5 = np.exp(-lam_c*(1/(tau*Tbstar+1)))
f1[5], f5[5]
```

```python
f6 = np.exp(-lam_c*(1/(tau*Tbstar+1)) + lam_c) / np.exp(lam_c)
# assert np.allclose(f1, f6)
f1[5], f6[5]
```

```python
f7 = np.exp(-lam_c*(1/(tau*Tbstar+1) - 1)) / np.exp(lam_c)
f1[5], f7[5]
```

```python
f8 = np.exp(lam_c*tau*Tbstar)/ np.exp(lam_c)
f1[5], f8[5]
```

```python
plt.plot(T, f7, label = "full expoential expression")
plt.plot(T, f8, label = 'linearization of the arguement in the exponent')
plt.xlabel('temperature, $T$ [K]')
plt.legend()
```

```python
plt.plot(T, (f7-f8)/f7)
plt.ylabel("fractional difference ")
plt.title("fractional difference between full and approximate exp expression")
plt.xlabel('temperature, $T$ [K]')
```
