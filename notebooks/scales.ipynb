```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
Ec = 60e3
R = 8.314
# Tavg = 263
```

```python
T = np.arange(253, 273, 1)
```

```python
Tavg = T.mean()

lam_c = Ec/R/Tavg
G = 50e-3
kappa0 = 2.1
z0 = 50
Tb0 = 10#G*z0/kappa0
tau = Tb0/Tavg
```

```python
Tb = T-Tavg
f1 = np.exp(-Ec/R/T)
f2 = np.exp(-Ec/R/(Tb+Tavg))
assert np.all(f1==f2)
```

```python
f3= np.exp(-Ec/R/Tavg*(1/(Tb/Tavg+1)))
assert np.allclose(f1, f3)
```

```python

Tbstar = Tb/Tb0
f4= np.exp(-Ec/R/Tavg*(1/(Tb0*Tbstar/Tavg+1)))
assert np.allclose(f1, f4)
```

```python
lam_c = Ec/R/Tavg
tau = Tb0/Tavg
f5 = np.exp(-lam_c*(1/(tau*Tbstar+1)))
assert np.allclose(f1, f5)
```

```python
f6 = np.exp(-lam_c*(1-tau*Tbstar))
```

```python
plt.plot(T, f5, label = "full expoential expression")
plt.plot(T, f6, label = 'linearization of the arguement in the exponent')
plt.xlabel('temperature, $T$ [K]')
plt.legend()
```

```python
plt.plot(T, (f5-f6)/f5)
plt.ylabel("fractional difference ")
plt.title("fractional difference between full and approximate exp expression")
plt.xlabel('temperature, $T$ [K]')
```

```python
f7 = np.exp(lam_c*tau*Tbstar)/np.exp(lam_c)
assert np.allclose(f6, f7)
plt.plot(Tb,np.exp(lam_c*tau*Tbstar))
plt.title('just the non-dimensional exp term')
plt.xlabel('difference temperature, $T$ [K]')
```

```python
f7 = np.exp(lam_c*tau*Tbstar)/np.exp(lam_c)
assert np.allclose(f6, f7)
plt.plot(Tb/Tb0,np.exp(lam_c*tau*Tbstar))
plt.title('just the non-dimensional exp term')
plt.xlabel('nondimensional difference temperature, $T$ [K]')
```

```python
print(f'lam_c: {lam_c}')
print(f'tau: {tau}')
print(f'lam_c * tau: {lam_c*tau}')
```
